```python
import gradio as gr
from pypdf import PdfReader

from langchain_google_genai import ChatGoogleGenerativeAI

reader = PdfReader("Resturaunt Q&A.pdf")
resturaunt_info  = ""
for page in reader.pages:
    text = page.extract_text()

    if text:
      resturaunt_info += text + "\n"
```

```python
def chatbot(message, history):
  prompt = f"""
  You are a restaurant customer support asssistant
  Use the following reataurant information
  to answer the questions asked by the user

  Resaurant Information:
  {restaurant_info}

  Customer Question:
  {message}

  Answer the customer clearly and politely
  If the information is not available in the
  restaurant information say "I don't know"
  """

  response = llm.invoke(prompt)
  return response.content

```

```python
demo = gr.ChatInterface(
      fn = chatbot,
      title="Restaurnt Support Bot",
      description="Ask questions about our restaurant"
  )
demo.launch()
```

```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.agents import create_agent
```

```python
loader = PyPDFLoader("/content/Resume(changed).pdf")
documents = loader.load()
```

```python
@tool
def read_pdf(question: str) -> str:
  """Answer a question using the pdf content """
  text = "\n".join(doc.page_content for doc in documents)
  return text
```

```python
agent = create_agent(
    model=llm,
    tools=[read_pdf],
    system_prompt="""
    You are a PDF reader agent.
    Use the read_pdf tool to find information from the
    answer only using information from the pdf.
    If the answer is not available in the pdf, say "I don't know.
    I could not find this information in the PDF"
    """,
)
```

```python
from langchain_core.tools import tool
```

```python
question = input("Ask a question")
result = agent.invoke({
    "messages": [
        {"role": "user", "content": question}
    ]
})

answer = result["messages"][-1].content

if isinstance(answer,list):
  answer = "\n".join(
      item["text"]
      for item in answer
      if isinstance(item, dict) and item.get("type") == "text"
  )
  print("\n" + "=" * 60)
  print("PDF agnet answered")
  print("=" * 60)
  print(answer)
```
